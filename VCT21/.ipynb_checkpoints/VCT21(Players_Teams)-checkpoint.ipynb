```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Loading data
players21 = pd.read_csv("../../../.cache/kagglehub/datasets/ryanluong1/valorant-champion-tour-2021-2023-data/versions/47/vct_2021/ids/players_ids.csv")
teams21 = pd.read_csv("../../../.cache/kagglehub/datasets/ryanluong1/valorant-champion-tour-2021-2023-data/versions/47/vct_2021/ids/teams_ids.csv")
players22 = pd.read_csv("../../../.cache/kagglehub/datasets/ryanluong1/valorant-champion-tour-2021-2023-data/versions/47/vct_2022/ids/players_ids.csv")
teams22 = pd.read_csv("../../../.cache/kagglehub/datasets/ryanluong1/valorant-champion-tour-2021-2023-data/versions/47/vct_2022/ids/teams_ids.csv")
players23 = pd.read_csv("../../../.cache/kagglehub/datasets/ryanluong1/valorant-champion-tour-2021-2023-data/versions/47/vct_2023/ids/players_ids.csv")
teams23 = pd.read_csv("../../../.cache/kagglehub/datasets/ryanluong1/valorant-champion-tour-2021-2023-data/versions/47/vct_2023/ids/teams_ids.csv")
players24 = pd.read_csv("../../../.cache/kagglehub/datasets/ryanluong1/valorant-champion-tour-2021-2023-data/versions/47/vct_2024/ids/players_ids.csv")
teams24 = pd.read_csv("../../../.cache/kagglehub/datasets/ryanluong1/valorant-champion-tour-2021-2023-data/versions/47/vct_2024/ids/teams_ids.csv")
players25 = pd.read_csv("../../../.cache/kagglehub/datasets//ryanluong1/valorant-champion-tour-2021-2023-data/versions/47/vct_2025/ids/players_ids.csv")
teams25 = pd.read_csv("../../../.cache/kagglehub/datasets//ryanluong1/valorant-champion-tour-2021-2023-data/versions/47/vct_2025/ids/teams_ids.csv")
players26 = pd.read_csv("../../../.cache/kagglehub/datasets//ryanluong1/valorant-champion-tour-2021-2023-data/versions/47/vct_2026/ids/players_ids.csv")
teams26 = pd.read_csv("../../../.cache/kagglehub/datasets//ryanluong1/valorant-champion-tour-2021-2023-data/versions/47/vct_2026/ids/teams_ids.csv")
```

```python
# Renaming columns
players21.rename(columns={"Player ID":"Player_ID"}, inplace = True)
teams21.rename(columns={"Team ID":"Team_ID"}, inplace = True)
players22.rename(columns={"Player ID":"Player_ID"}, inplace = True)
teams22.rename(columns={"Team ID":"Team_ID"}, inplace = True)
players23.rename(columns={"Player ID":"Player_ID"}, inplace = True)
teams23.rename(columns={"Team ID":"Team_ID"}, inplace = True)
players24.rename(columns={"Player ID":"Player_ID"}, inplace = True)
teams24.rename(columns={"Team ID":"Team_ID"}, inplace = True)
players25.rename(columns={"Player ID":"Player_ID"}, inplace = True)
teams25.rename(columns={"Team ID":"Team_ID"}, inplace = True)
players26.rename(columns={"Player ID":"Player_ID"}, inplace = True)
teams26.rename(columns={"Team ID":"Team_ID"}, inplace = True)
```

```python
# arranging in ascending order
players21.sort_values(by="Player_ID", ascending=True, inplace=True)
teams21.sort_values(by="Team_ID", ascending=True, inplace=True)
players22.sort_values(by="Player_ID", ascending=True, inplace=True)
teams22.sort_values(by="Team_ID", ascending=True, inplace=True)
players23.sort_values(by="Player_ID", ascending=True, inplace=True)
teams23.sort_values(by="Team_ID", ascending=True, inplace=True)
players24.sort_values(by="Player_ID", ascending=True, inplace=True)
teams24.sort_values(by="Team_ID", ascending=True, inplace=True)
players25.sort_values(by="Player_ID", ascending=True, inplace=True)
teams25.sort_values(by="Team_ID", ascending=True, inplace=True)
players26.sort_values(by="Player_ID", ascending=True, inplace=True)
teams26.sort_values(by="Team_ID", ascending=True, inplace=True)
```

```python
# dropping null values
players21 = players21.dropna()
players22 = players22.dropna()
players23 = players23.dropna()
players24 = players24.dropna()
players25 = players25.dropna()
players26 = players26.dropna()
teams21 = teams21.dropna()
teams22 = teams22.dropna()
teams23 = teams23.dropna()
teams24 = teams24.dropna()
teams25 = teams25.dropna()
teams26 = teams26.dropna()
```

```python
players = (
    players21
    .merge(players22, on="Player", how="outer", suffixes=('', '_df2'))
    .merge(players23, on="Player", how="outer", suffixes=('', '_df3'))
    .merge(players24, on="Player", how="outer", suffixes=('', '_df4'))
    .merge(players25, on="Player", how="outer", suffixes=('', '_df4'))
    .merge(players26, on="Player", how="outer", suffixes=('', '_df4'))
)
players.sort_values(by="Player_ID", ascending=True, inplace=True)
players["ID"] = players[['Player_ID_df2', 'Player_ID', 'Player_ID_df3', 'Player_ID_df4','Player_ID_df5','Player_ID_df6']].bfill(axis=1).iloc[:, 0]
players = players.drop(columns=['Player_ID','Player_ID_df2','Player_ID_df3','Player_ID_df4','Player_ID_df5','Player_ID_df6'])


teams = (
    teams21
    .merge(teams22, on="Team", how="outer", suffixes=('', '_df2'))
    .merge(teams23, on="Team", how="outer", suffixes=('', '_df3'))
    .merge(teams24, on="Team", how="outer", suffixes=('', '_df4'))
    .merge(teams25, on="Team", how="outer", suffixes=('', '_df4'))
    .merge(teams26, on="Team", how="outer", suffixes=('', '_df4'))
)
teams.sort_values(by="Team_ID", ascending=True, inplace=True)
teams["ID"] = teams[['Team_ID_df2', 'Team_ID', 'Team_ID_df3', 'Team_ID_df4','Team_ID_df5','Team_ID_df6']].bfill(axis=1).iloc[:, 0]
teams = teams.drop(columns=['Team_ID','Team_ID_df2','Team_ID_df3','Team_ID_df4','Team_ID_df5','Team_ID_df6'])
```
