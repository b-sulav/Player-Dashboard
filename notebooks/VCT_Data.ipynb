```python
# Dropping unnecessary fields
stats21.drop(columns=["Stage", "Match Type"], inplace=True)
stats22.drop(columns=["Stage", "Match Type"], inplace=True)
stats23.drop(columns=["Stage", "Match Type"], inplace=True)
stats24.drop(columns=["Stage", "Match Type"], inplace=True)
stats25.drop(columns=["Stage", "Match Type"], inplace=True)
stats26.drop(columns=["Stage", "Match Type"], inplace=True)
```

```python
# Typecasting string to integer for mean calculation
stats21["Headshot %"] = stats21["Headshot %"].str.replace("%","", regex=False).astype('float')
stats22["Headshot %"] = stats22["Headshot %"].str.replace("%","", regex=False).astype('float')
stats23["Headshot %"] = stats23["Headshot %"].str.replace("%","", regex=False).astype('float')
stats24["Headshot %"] = stats24["Headshot %"].str.replace("%","", regex=False).astype('float')
stats25["Headshot %"] = stats25["Headshot %"].str.replace("%","", regex=False).astype('float')
stats26["Headshot %"] = stats26["Headshot %"].str.replace("%","", regex=False).astype('float')

stats21["Clutch Success %"] = stats21["Clutch Success %"].str.replace("%","", regex=False).astype('float')
stats22["Clutch Success %"] = stats22["Clutch Success %"].str.replace("%","", regex=False).astype('float')
stats23["Clutch Success %"] = stats23["Clutch Success %"].str.replace("%","", regex=False).astype('float')
stats24["Clutch Success %"] = stats24["Clutch Success %"].str.replace("%","", regex=False).astype('float')
stats25["Clutch Success %"] = stats25["Clutch Success %"].str.replace("%","", regex=False).astype('float')
stats26["Clutch Success %"] = stats26["Clutch Success %"].str.replace("%","", regex=False).astype('float')
```

```python
# Filling null values with 0
playersstats21 = playersstats21.fillna(0)
playersstats22 = playersstats22.fillna(0)
playersstats23 = playersstats23.fillna(0)
playersstats24 = playersstats24.fillna(0)
playersstats25 = playersstats25.fillna(0)
playersstats26 = playersstats26.fillna(0)
```

```python
# Calculating the Percentile for each scoring catergories
playersstats21["ACS Percentile"] = playersstats21["Average Combat Score"].rank(pct=True)*100 
playersstats21["K/D Percentile"] = playersstats21["Kills:Deaths"].rank(pct=True)*100 
playersstats21["ADR Percentile"] = playersstats21["Average Damage Per Round"].rank(pct=True)*100 
playersstats21["Headshot Percentile"] = playersstats21["Headshot %"].rank(pct=True)*100 

playersstats22["ACS Percentile"] = playersstats22["Average Combat Score"].rank(pct=True)*100 
playersstats22["K/D Percentile"] = playersstats22["Kills:Deaths"].rank(pct=True)*100 
playersstats22["ADR Percentile"] = playersstats22["Average Damage Per Round"].rank(pct=True)*100 
playersstats22["Headshot Percentile"] = playersstats22["Headshot %"].rank(pct=True)*100 

playersstats23["ACS Percentile"] = playersstats23["Average Combat Score"].rank(pct=True)*100 
playersstats23["K/D Percentile"] = playersstats23["Kills:Deaths"].rank(pct=True)*100 
playersstats23["ADR Percentile"] = playersstats23["Average Damage Per Round"].rank(pct=True)*100 
playersstats23["Headshot Percentile"] = playersstats23["Headshot %"].rank(pct=True)*100 

playersstats24["ACS Percentile"] = playersstats24["Average Combat Score"].rank(pct=True)*100 
playersstats24["K/D Percentile"] = playersstats24["Kills:Deaths"].rank(pct=True)*100 
playersstats24["ADR Percentile"] = playersstats24["Average Damage Per Round"].rank(pct=True)*100 
playersstats24["Headshot Percentile"] = playersstats24["Headshot %"].rank(pct=True)*100 

playersstats25["ACS Percentile"] = playersstats25["Average Combat Score"].rank(pct=True)*100 
playersstats25["K/D Percentile"] = playersstats25["Kills:Deaths"].rank(pct=True)*100 
playersstats25["ADR Percentile"] = playersstats25["Average Damage Per Round"].rank(pct=True)*100 
playersstats25["Headshot Percentile"] = playersstats25["Headshot %"].rank(pct=True)*100 

playersstats26["ACS Percentile"] = playersstats26["Average Combat Score"].rank(pct=True)*100 
playersstats26["K/D Percentile"] = playersstats26["Kills:Deaths"].rank(pct=True)*100 
playersstats26["ADR Percentile"] = playersstats26["Average Damage Per Round"].rank(pct=True)*100 
playersstats26["Headshot Percentile"] = playersstats26["Headshot %"].rank(pct=True)*100 
```

```python
# Calculating the Player Performance Score (in Percentile)
playersstats21["Average Percentile"] = (playersstats21[["ACS Percentile" , "K/D Percentile" , "ADR Percentile" , "Headshot Percentile"]].mean(axis=1)).rank(pct=True)*100
playersstats22["Average Percentile"] = (playersstats22[["ACS Percentile" , "K/D Percentile" , "ADR Percentile" , "Headshot Percentile"]].mean(axis=1)).rank(pct=True)*100
playersstats23["Average Percentile"] = (playersstats23[["ACS Percentile" , "K/D Percentile" , "ADR Percentile" , "Headshot Percentile"]].mean(axis=1)).rank(pct=True)*100
playersstats24["Average Percentile"] = (playersstats24[["ACS Percentile" , "K/D Percentile" , "ADR Percentile" , "Headshot Percentile"]].mean(axis=1)).rank(pct=True)*100
playersstats25["Average Percentile"] = (playersstats25[["ACS Percentile" , "K/D Percentile" , "ADR Percentile" , "Headshot Percentile"]].mean(axis=1)).rank(pct=True)*100
playersstats26["Average Percentile"] = (playersstats26[["ACS Percentile" , "K/D Percentile" , "ADR Percentile" , "Headshot Percentile"]].mean(axis=1)).rank(pct=True)*100
```
